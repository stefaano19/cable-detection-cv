```python
import cv2
import numpy as np

def thinning(mask):
    """
    mask: np.uint8 {0,1}
    return: skeleton np.uint8 {0,1}
    """
    mask = (mask > 0).astype(np.uint8) * 255

    skel = np.zeros(mask.shape, np.uint8)
    element = cv2.getStructuringElement(cv2.MORPH_CROSS, (3,3))

    done = False
    while not done:
        eroded = cv2.erode(mask, element)
        temp = cv2.dilate(eroded, element)
        temp = cv2.subtract(mask, temp)
        skel = cv2.bitwise_or(skel, temp)
        mask = eroded.copy()

        if cv2.countNonZero(mask) == 0:
            done = True

    return (skel > 0).astype(np.uint8)
```

```python
def estimate_rho_theta_from_mask(mask):
    edges = (mask * 255).astype(np.uint8)

    lines = cv2.HoughLines(
        edges,
        rho=1,
        theta=np.pi / 180,
        threshold=30
    )

    if lines is None:
        return 0.0, 0.0

    rhos = []
    sin_t = []
    cos_t = []

    for l in lines[:10]:
        rho, theta = l[0]
        rhos.append(rho)
        sin_t.append(np.sin(theta))
        cos_t.append(np.cos(theta))

    rho_mean = float(np.mean(rhos))
    theta_mean = float(np.arctan2(np.mean(sin_t), np.mean(cos_t)))

    if theta_mean < 0:
        theta_mean += np.pi

    return rho_mean, theta_mean
```

```python
def select_dominant_line(preds, angle_eps=np.deg2rad(5)):
    if len(preds) == 1:
        return preds

    clusters = []

    for p in preds:
        placed = False
        for c in clusters:
            if abs(p["theta"] - c[0]["theta"]) < angle_eps:
                c.append(p)
                placed = True
                break
        if not placed:
            clusters.append([p])

    # cluster dominante = più elementi
    dominant = max(clusters, key=len)

    # tieni la linea con score massimo
    return [max(dominant, key=lambda x: x["score"])]
```

```python
def mask_to_bbox_area(mask):
    ys, xs = np.where(mask > 0)

    if len(xs) == 0:
        return [0, 0, 0, 0], 0

    x0, x1 = xs.min(), xs.max()
    y0, y1 = ys.min(), ys.max()

    w = max(1, x1 - x0 + 1)
    h = max(1, y1 - y0 + 1)
    area = int(mask.sum())

    return [int(x0), int(y0), int(w), int(h)], area
```

```python
from pycocotools import mask as mask_utils

def mask_to_rle(mask):
    """
    mask: np.uint8 {0,1}
    return: RLE dict COCO
    """

    rle = mask_utils.encode(
        np.asfortranarray(mask.astype(np.uint8))
    )

    # COCO richiede stringa
    rle["counts"] = rle["counts"].decode("ascii")
    return rle
```

```python
def run_validation_pipeline(model, val_loader, epoch):
    model.eval()
    predictions = []

    with torch.no_grad():
        for batch in val_loader:
            if batch is None:
                continue

            images, targets = batch
            img = images[0].to(device)
            target = targets[0]
            image_id = int(target["image_id"].item())

            output = model([img])[0]

            labels = output["labels"].cpu().numpy()
            scores = output["scores"].cpu().numpy()
            masks  = output["masks"].cpu().numpy()

            SCORE_TH = 0.01   # 🔥 PIÙ BASSA IN VALIDATION
            keep = np.where((labels == 1) & (scores >= SCORE_TH))[0]

            for k in keep:
                mask = (masks[k, 0] > 0.5).astype(np.uint8)

                area = mask.sum()
                if area < 50:        # 🔧 PIÙ PERMISSIVO
                    continue

                ys, xs = np.where(mask)
                if len(xs) == 0:
                    continue

                h = ys.max() - ys.min()
                w = xs.max() - xs.min()
                elong = max(h, w) / (min(h, w) + 1e-6)
                """if elong < 2.0:      # 🔧 PIÙ PERMISSIVO
                    continue"""

                thin = thinning(mask)
                rho, theta = estimate_rho_theta_from_mask(thin if thin.sum() > 20 else mask)

                bbox, _ = mask_to_bbox_area(mask)
                rle = mask_to_rle(mask)

                predictions.append({
                    "image_id": image_id,
                    "category_id": 0,
                    "segmentation": rle,
                    "bbox": bbox,
                    "score": float(scores[k]),
                    "lines": [float(rho), float(theta)]
                })

    if len(predictions) == 0:
        print(f"⚠️ Epoch {epoch}: nessuna predizione in validation → skip LDS")
        return None

    pred_path = f"/kaggle/working/val_preds_epoch_{epoch}.json"
    with open(pred_path, "w") as f:
        json.dump(predictions, f)

    return pred_path
```

```python
from torchvision.datasets import CocoDetection
from pycocotools import mask as coco_mask
from PIL import Image

class CableDataset(CocoDetection):
    def __init__(self, img_folder, ann_file, transforms=None):
        super().__init__(img_folder, ann_file)
        self._transforms = transforms

    def __getitem__(self, idx):
        img, anns = super().__getitem__(idx)
        w, h = img.size

        boxes = []
        labels = []
        masks = []

        for ann in anns:
            if "segmentation" not in ann:
                continue

            x, y, bw, bh = ann["bbox"]

            # 🔥 FILTRO BOX INVALIDI
            if bw <= 1 or bh <= 1:
                continue

            x2 = x + bw
            y2 = y + bh

            # sicurezza floating point
            if x2 <= x or y2 <= y:
                continue

            boxes.append([x, y, x2, y2])
            labels.append(1)


            # segmentation -> mask
            rles = coco_mask.frPyObjects(
                ann["segmentation"], h, w
            )
            m = coco_mask.decode(rles)
            if m.ndim == 3:
                m = np.any(m, axis=2)

            masks.append(m.astype(np.uint8))

        # 🔴 CONTROLLO CORRETTO: DOPO il loop
        if len(boxes) == 0:
            return self.__getitem__((idx + 1) % len(self))

        target = {
            "boxes": torch.as_tensor(boxes, dtype=torch.float32),
            "labels": torch.as_tensor(labels, dtype=torch.int64),
            "masks": torch.as_tensor(np.array(masks), dtype=torch.uint8),
            "image_id": torch.tensor([self.ids[idx]]),
        }

        if self._transforms:
            img = self._transforms(img)

        return img, target
```

```python
from pycocotools.coco import COCO
from pycocotools.cocoeval import COCOeval
from pathlib import Path
from collections import defaultdict
from tqdm import tqdm
from pycocotools import mask as coco_mask
import json
import numpy as np
import cv2
```

```python
def evaluate_segmentation(gt_json_path, pred_json_path, check_cable_class=False):
    # Load ground truth
    coco_gt = COCO(gt_json_path)

    # Load predictions
    with open(pred_json_path, 'r') as f:
        predictions = json.load(f)

    # Load results into COCO results structure
    coco_res = coco_gt.loadRes(predictions)

    # Create COCOeval object
    coco_eval = COCOeval(coco_gt, coco_res, 'segm')
    if check_cable_class:
        coco_eval.params.catIds = [1]  # id of the cable class

    # Run evaluation
    coco_eval.evaluate()
    coco_eval.accumulate()
    coco_eval.summarize()

    avg_p50 = coco_eval.stats[1]
    avg_r50 = coco_eval.stats[7]
    return avg_p50, avg_r50
```

```python
def combined_analysis(gt_annotation_file, prediction_file):
    # Load ground truth data
    with open(gt_annotation_file, 'r') as f:
        gt_data = json.load(f)
    # Load prediction data
    with open(prediction_file, 'r') as f:
        pred_data = json.load(f)
    # Group GT lines by image id
    gt_lines_by_image = defaultdict(list)
    for ann in gt_data['annotations']:
        image_id = ann['image_id']
        if 'polar_coordinates' in ann:
            lines = [(coord['rho'], coord['theta']) for coord in ann['polar_coordinates']]
            gt_lines_by_image[image_id].extend(lines)
        else:
            raise RuntimeError(f'no polar coord for image id {image_id}')
    # Group predictions by image id
    pred_by_image = defaultdict(list)
    for pred in pred_data:
        pred_by_image[pred['image_id']].append(pred)
    angle_diffs = []
    rho_diffs = []

    def theta_diff(theta_pred, theta_gt):
        t = min(abs(theta_pred - theta_gt), np.pi - abs(theta_pred - theta_gt))
        return np.exp(-.12 * t)

    def polygons_to_mask(polygons, shape):
        mask = np.zeros(shape, dtype=np.uint8)
        for polygon in polygons:
            pts = np.array(polygon).reshape((-1, 2)).astype(np.int32)
            cv2.fillPoly(mask, [pts], color=255)
        return mask

    def compute_iou(mask1, mask2):
        intersection = np.logical_and(mask1, mask2).sum()
        union = np.logical_or(mask1, mask2).sum()
        return intersection / union if union > 0 else 0

    total_matches = 0
    total_gt_lines = 0
    total_pred_lines = 0

    for image_info in tqdm(gt_data['images']):
        image_id = image_info['id']
        height, width = image_info['height'], image_info['width']

        # Load predictions for this image
        pred_masks = []
        pred_lines = []
        for pred in pred_by_image.get(image_id, []):
            seg = pred['segmentation']
            if isinstance(seg, list):
                mask_poly = polygons_to_mask(seg, (height, width))
                pred_masks.append(mask_poly)
            elif isinstance(seg, dict) and 'counts' in seg and 'size' in seg:
                mask_rle = coco_mask.decode(seg)
                if mask_rle.ndim == 3:
                    mask_rle = mask_rle[:, :, 0]
                mask_rle = (mask_rle * 255).astype(np.uint8)
                pred_masks.append(mask_rle)
            else:
                raise RuntimeError(f'[SEGM] unsupported format for image id {image_id}')

            # Extract predicted line if exists
            if 'lines' in pred and len(pred['lines']) == 2:
                rho, theta = pred['lines']
                rho = np.abs(rho / np.sqrt(height**2 + width**2))
                pred_lines.append((rho, theta))
            else:
                pred_lines.append(None)

        # Load ground truth masks for this image
        gt_masks = []
        gt_lines = []
        for ann in gt_data['annotations']:
            if ann['image_id'] == image_id:
                seg = ann['segmentation']
                if isinstance(seg, list):
                    mask_poly = polygons_to_mask(seg, (height, width))
                    gt_masks.append(mask_poly)
                elif isinstance(seg, dict) and 'counts' in seg and 'size' in seg:
                    mask_rle = coco_mask.decode(seg)
                    if mask_rle.ndim == 3:
                        mask_rle = mask_rle[:, :, 0]
                    mask_rle = (mask_rle * 255).astype(np.uint8)
                    gt_masks.append(mask_rle)
                else:
                    raise RuntimeError(f'[GT] unsupported format for image id {image_id}')

                # Extract GT line
                if 'polar_coordinates' in ann and len(ann['polar_coordinates']) > 0:
                    rho, theta = ann['polar_coordinates'][0]['rho'], ann['polar_coordinates'][0]['theta']
                    rho = np.abs(rho / np.sqrt(height**2 + width**2))
                    gt_lines.append((rho, theta))
                else:
                    gt_lines.append(None)

        # Detect the matching mask by IoU
        matched_gt = set()
        for pred_idx, pred_mask in enumerate(pred_masks):
            best_iou = 0
            best_gt_idx = -1

            for gt_idx, gt_mask in enumerate(gt_masks):
                if gt_idx in matched_gt:
                    continue
                iou = compute_iou(pred_mask, gt_mask)
                if iou > best_iou:
                    best_iou = iou
                    best_gt_idx = gt_idx

            # Consider it a match if IoU > threshold (e.g., 0.5)
            #if best_iou > 0.5:
            if best_gt_idx >= 0:
                matched_gt.add(best_gt_idx)
                total_matches += 1

                # Compute the rho_diff and theta_diff if both lines exist
                pred_line = pred_lines[pred_idx]
                gt_line = gt_lines[best_gt_idx]

                if pred_line is not None and gt_line is not None:
                    rho_pred, theta_pred = pred_line
                    rho_gt, theta_gt = gt_line

                    rho_diffs.append(abs(rho_pred - rho_gt))
                    angle_diffs.append(theta_diff(theta_pred, theta_gt))

        # Count matching and not matching lines
        total_gt_lines += len(gt_masks)
        total_pred_lines += len(pred_masks)

    print(f"Total GT lines: {total_gt_lines}")
    print(f"Total predicted lines: {total_pred_lines}")
    print(f"Total matches: {total_matches}")
    print(f"Lines with coordinate differences computed: {len(rho_diffs)}")

    if len(rho_diffs) == 0:
        return 0, 0

    return np.mean(rho_diffs), np.mean(angle_diffs)
```

```python
def compute_line_detection_score(gt_json_path, pred_json_path):

    avg_p50, avg_r50 = evaluate_segmentation(gt_json_path, pred_json_path)
    rho_diff, angle_diff = combined_analysis(gt_json_path, pred_json_path)

    print(f'{avg_p50=}, {avg_r50=}, {angle_diff=}')

    lds = avg_p50 + avg_r50 + 2 * angle_diff
    print(f'LDS = {lds}')
    return lds
```

```python
import json
import os

def patch_gt_for_coco(gt_json_path, out_path):
    # Verifica che il file di input esista
    if not os.path.exists(gt_json_path):
        print(f"❌ Errore: Il file {gt_json_path} non esiste.")
        return None

    with open(gt_json_path, "r") as f:
        gt = json.load(f)

    # Aggiunge le chiavi mancanti per il formato COCO
    gt.setdefault("info", {
        "description": "Dataset patched for COCO",
        "version": "1.0",
        "year": 2024,
        "contributor": "",
        "date_created": ""
    })
    gt.setdefault("licenses", [])
    
    # Nota: Assicurati che l'ID 0 sia corretto per il tuo modello. 
    # Spesso in COCO gli ID partono da 1 e 0 è il background, ma dipende dalla libreria che usi.
    gt.setdefault("categories", [
    {"id": 1, "name": "cable"}  # ERA 0, DEVE ESSERE 1
    ])

    # Salva il file modificato
    with open(out_path, "w") as f:
        json.dump(gt, f)

    print(f"✅ JSON patchato salvato in: {out_path}")
    return out_path
```

```python
from torchvision.models.detection import maskrcnn_resnet50_fpn

model = maskrcnn_resnet50_fpn(weights="DEFAULT")

# 2 classi: background + cable
in_features = model.roi_heads.box_predictor.cls_score.in_features
model.roi_heads.box_predictor = torchvision.models.detection.faster_rcnn.FastRCNNPredictor(
    in_features, 2
)

in_features_mask = model.roi_heads.mask_predictor.conv5_mask.in_channels
model.roi_heads.mask_predictor = torchvision.models.detection.mask_rcnn.MaskRCNNPredictor(
    in_features_mask, 256, 2
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
```

```python
params = [p for p in model.parameters() if p.requires_grad]
optimizer = torch.optim.AdamW(params, lr=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",        # perché ottimizzi LDS
    factor=0.5,
    patience=3,
    min_lr=1e-6,
    verbose=True
)
```

```python
start_epoch = 0
best_lds = -1.0
patience = 10
no_improve = 0

if os.path.exists(CKPT_PATH):
    ckpt = torch.load(
        CKPT_PATH,
        map_location=device,
        weights_only=False   # 🔥 FIX
    )

    model.load_state_dict(ckpt["model_state"])
    optimizer.load_state_dict(ckpt["optimizer_state"])

    start_epoch = ckpt.get("epoch", -1) + 1
    best_lds = ckpt.get("best_lds", -1.0)

    print(f"🔁 Resume da epoca {start_epoch}, best LDS={best_lds:.3f}")
else:
    print("🆕 Training da zero")
```

```python
train_transform_strong = T.Compose([
    T.RandomHorizontalFlip(0.5),
    T.RandomVerticalFlip(0.5),
    T.RandomRotation(5),
    T.ColorJitter(0.15, 0.15, 0.1, 0),
    T.ToTensor(),
])
```

```python
train_transform_light = T.Compose([
    T.RandomHorizontalFlip(0.5),
    T.RandomRotation(2),
    T.ToTensor(),
])
```

```python
train_transform_off = T.Compose([
    T.ToTensor(),
])
```

```python
NUM_EPOCHS = 80
VAL_EVERY = 1
AUG_OFF_EPOCH = 55
AUG_LIGHT_EPOCH = 30


for epoch in range(start_epoch, NUM_EPOCHS):

    if epoch < AUG_LIGHT_EPOCH:
        full_dataset._transforms = train_transform_strong
    elif epoch < AUG_OFF_EPOCH:
        full_dataset._transforms = train_transform_light
    else:
        full_dataset._transforms = train_transform_off
    model.train()
    epoch_loss = 0.
    valid_batches = 0


    # ===== TRAIN =====
    for batch in tqdm(train_loader, desc=f"Epoch {epoch}"):

        if batch is None:
            continue   # 🔥 batch vuoto → skip
        valid_batches += 1
        images, targets = batch
    
        images = [img.to(device) for img in images]
        targets = [{k: v.to(device) for k, v in t.items()} for t in targets]
    
        loss_dict = model(images, targets)
        losses = sum(loss for loss in loss_dict.values())
    
        optimizer.zero_grad()
        losses.backward()
        optimizer.step()
    
        epoch_loss += losses.item()


    print(f"Epoch {epoch} | Loss: {epoch_loss:.4f} | valid batches = {valid_batches}")

    # ===== VALIDATION + LDS =====
    if epoch % VAL_EVERY == 0:
        full_dataset._transforms = test_transform
        val_pred_path = run_validation_pipeline(
            model, val_loader, epoch
        )

        if val_pred_path is None:
            print("⏭️ LDS saltato (nessuna predizione)")
            continue
        
        lds = compute_line_detection_score(GT_VAL_JSON, val_pred_path)


        # scheduler SOLO se è ReduceLROnPlateau
        scheduler.step(lds)

        print(f"📊 Epoch {epoch} | LDS = {lds:.4f}")

        # ===== EARLY STOPPING =====
        if lds > best_lds:
            best_lds = lds
            no_improve = 0

            torch.save({
                "epoch": epoch,
                "model_state": model.state_dict(),
                "optimizer_state": optimizer.state_dict(),
                "best_lds": float(best_lds)
            }, CKPT_PATH)

            torch.save(model.state_dict(), BEST_MODEL_PATH)

            print("💾 Best model salvato")
        else:
            no_improve += 1
            print(f"⚠️ No improve ({no_improve}/{patience})")

        if no_improve >= patience:
            print("🛑 Early stopping attivato")
            break
```

```python
import cv2
import json
import numpy as np
from PIL import Image
from tqdm import tqdm

predictions = []

# ===== PARAMETRI FINALI =====
SCORE_TH = 0.01
MIN_AREA = 30
MIN_ELONG = 1.5      # forma allungata (cavo)
ANGLE_EPS = np.deg2rad(5)  # clustering angolare (±5°)

def angle_dist(a, b):
    d = abs(a - b)
    return min(d, np.pi - d)

with torch.no_grad():
    for fname in tqdm(test_images):

        image_id = name2id[fname]

        img_path = f"{ROOT}/test/{fname}"
        img = Image.open(img_path).convert("RGB")
        tensor = test_transform(img).to(device)

        output = model([tensor])[0]

        labels = output["labels"].cpu().numpy()
        scores = output["scores"].cpu().numpy()
        masks  = output["masks"].cpu().numpy()

        # 1️⃣ selezione iniziale (classe + score)
        keep = np.where((labels == 1) & (scores >= SCORE_TH))[0]
        if len(keep) == 0:
            continue

        image_preds = []   # ⬅ raccoglie TUTTE le linee candidate

        for idx in keep:
            mask = (masks[idx, 0] > 0.5).astype(np.uint8)
            score = float(scores[idx])

            # 2️⃣ filtro area
            area = mask.sum()
            if area < MIN_AREA:
                continue

            # 3️⃣ filtro elongazione
            ys, xs = np.where(mask)
            if len(xs) == 0:
                continue

            h = ys.max() - ys.min()
            w = xs.max() - xs.min()
            elong = max(h, w) / (min(h, w) + 1e-6)
            if elong < MIN_ELONG:
                continue

            # 4️⃣ thinning
            # 4️⃣ thinning
            thin = thinning(mask)
            
            # ===== NUOVO: filtra pixel troppo "ramificati" =====
            kernel = np.array([[1,1,1],
                               [1,10,1],
                               [1,1,1]], dtype=np.uint8)
            
            neighbor_count = cv2.filter2D(thin, -1, kernel)
            thin = np.where(neighbor_count <= 12, thin, 0).astype(np.uint8)
            
            # tieni solo componente principale
            nlab, lab, stats, _ = cv2.connectedComponentsWithStats(thin, connectivity=8)
            if nlab > 1:
                largest = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
                thin = (lab == largest).astype(np.uint8)


            # 5️⃣ stima linea
            if thin.sum() < 20:
                rho, theta = estimate_rho_theta_from_mask(mask)
            else:
                rho, theta = estimate_rho_theta_from_mask(thin)
            
            theta = float(np.clip(theta, 0, np.pi))


            # ===== DEBUG VISIVO (QUI) =====
            if DEBUG_VIS and debug_count < DEBUG_MAX:
                debug_count += 1
            
                vis = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)
            
                # overlay mask (verde)
                overlay = vis.copy()
                overlay[mask > 0] = (0, 255, 0)
                vis = cv2.addWeighted(vis, 0.7, overlay, 0.3, 0)
            
                # overlay skeleton (rosso)
                ys, xs = np.where(thin > 0)
                vis[ys, xs] = (0, 0, 255)
            
                # disegna linea stimata
                a, b = np.cos(theta), np.sin(theta)
                x0, y0 = a * rho, b * rho
                pt1 = (int(x0 + 1000 * (-b)), int(y0 + 1000 * a))
                pt2 = (int(x0 - 1000 * (-b)), int(y0 - 1000 * a))
                cv2.line(vis, pt1, pt2, (255, 0, 0), 2)
            
                plt.figure(figsize=(6, 6))
                plt.title(f"{fname} | score={score:.2f} | θ={np.degrees(theta):.1f}°")
                plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
                plt.axis("off")
                plt.show()
            # ===== FINE DEBUG =====

            bbox, _ = mask_to_bbox_area(mask)
            rle = mask_to_rle(mask)

            image_preds.append({
                "rho": rho,
                "theta": theta,
                "score": score,
                "rle": rle,
                "bbox": bbox
            })

        if len(image_preds) == 0:
            continue

        # 6️⃣ CLUSTER ANGOLARE (linea dominante)
        clusters = []
        for p in image_preds:
            placed = False
            for c in clusters:
                if angle_dist(p["theta"], c[0]["theta"]) < ANGLE_EPS:
                    c.append(p)
                    placed = True
                    break
            if not placed:
                clusters.append([p])

        dominant_cluster = max(clusters, key=len)

        # 7️⃣ scegli la migliore nel cluster dominante
        best = max(dominant_cluster, key=lambda x: x["score"])

        # 8️⃣ salva UNA predizione per immagine
        predictions.append({
            "image_id": image_id,
            "category_id": 0,
            "segmentation": best["rle"],
            "bbox": best["bbox"],
            "score": float(best["score"]),
            "lines": [float(best["rho"]), float(best["theta"])]
        })
```
